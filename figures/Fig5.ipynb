```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator

plt.rc('figure', facecolor='w')
plt.rc('font', size=11)
```

```python
def read_pmf(filename):
    pmf = np.loadtxt(filename).T
    pmf[1] -= np.min(pmf[1,:20])
    return pmf

pmf_wt = read_pmf('CHO/PMF/pmf_bs_wt.dat')
pmf_cit = read_pmf('CHO/PMF/pmf_bs_cit.dat')
```

```python
fig, ax = plt.subplots(figsize=(6,4))
ax.set_xlim([-2.6, 2.2])
ax.set_ylim([-24, 16])
ax.set_yticks(np.arange(-24, 17, 4))
ax.xaxis.set_minor_locator(MultipleLocator(0.2))
ax.yaxis.set_minor_locator(MultipleLocator(2))

ax.spines['top'].set_visible(False)
ax.spines['left'].set_position(('data', -2.7))
ax.spines['bottom'].set_position(('data', -25))
ax.spines['right'].set_visible(False)

ax.set_xlabel('RC ($\\AA$)')
ax.set_ylabel('$\\Delta G$ (kcal/mol)', labelpad=0)

ax.plot(*pmf_wt, label='$Bs$ChoM WT ($\\Delta^{\\ddagger} $G=11.8 kcal/mol)')
ax.plot(*pmf_cit, label='$Bs$ChoM:Arg90Cit ($\\Delta^{\\ddagger} $G=15.2 kcal/mol)')

for y in range(-24, 17, 4):
    ax.axhline(y=y, color='lightgrey', linewidth=0.5, zorder=-1, clip_on=False)

ax.legend(bbox_to_anchor=(0.77, 0.21), frameon=False)
plt.savefig('Fig5.png', dpi=300, bbox_inches='tight')
plt.show()
```
